## Пазл 1: Петухи и поросята

По тропинке вдоль кустов
Шли 11 хвостов.
Сосчитать я также смог,
Что шагало 30 ног.
Это вместе шли куда-то
Петухи и поросята.
А теперь вопрос таков:
Сколько было петухов?
И узнать я был бы рад,
Сколько было поросят?

````python
from z3 import *

roosters, pigs = Ints('roosters pigs')
solve(pigs + roosters == 11,        # 11 хвостов
      roosters * 2 + pigs * 4 == 30 # 30 ног
      )
````

## Пазл 2: Три новые задачки

### 2.1 Криптарифметика: SEND + MORE = MONEY

Каждая буква обозначает свою цифру (одинаковые буквы — одинаковые цифры,
разные буквы — разные цифры). Числа не начинаются с нуля. Найти, каким
цифрам соответствуют буквы, чтобы равенство

```
  S E N D
+ M O R E
---------
M O N E Y
```

выполнялось.

````python
S, E, N, D, M, O, R, Y = Ints('S E N D M O R Y')
digits = [S, E, N, D, M, O, R, Y]

solve(Distinct(*digits),
      *[And(d >= 0, d <= 9) for d in digits],
      S != 0, M != 0,
      (S*1000 + E*100 + N*10 + D) + (M*1000 + O*100 + R*10 + E)
      == (M*10000 + O*1000 + N*100 + E*10 + Y)
      )
````

### 2.2 Рыцари и лжецы

На острове живут рыцари (всегда говорят правду) и лжецы (всегда лгут).
Встретились трое: А, Б и В.

- А говорит: "Б и В — оба лжецы."
- Б говорит: "А — рыцарь."
- В говорит: "Я и А — разного типа."

Кто из них рыцарь, а кто лжец?

````python
A, B, C = Bools('A B C')  # True = рыцарь, False = лжец

# Житель говорит правду тогда и только тогда, когда он рыцарь
solve(A == And(Not(B), Not(C)),   # А: "Б и В — оба лжецы"
      B == A,                     # Б: "А — рыцарь"
      C == (C != A)               # В: "Я и А — разного типа"
      )
````

### 2.3 Ферма с курами, коровами и пауками

На ферме держат кур (2 ноги), коров (4 ноги) и пауков (8 ног).
Всего на ферме 20 голов и 100 ног. Известно, что коров вдвое больше,
чем пауков. Сколько на ферме кур, коров и пауков?

````python
chickens, cows, spiders = Ints('chickens cows spiders')

solve(chickens + cows + spiders == 20,               # 20 голов
      chickens*2 + cows*4 + spiders*8 == 100,         # 100 ног
      cows == 2 * spiders,                            # коров вдвое больше пауков
      chickens >= 0, cows >= 0, spiders >= 0
      )
````
